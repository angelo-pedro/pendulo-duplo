```python
import cv2
```

```python
video = cv2.VideoCapture(r"")

if not video.isOpened():
    raise RuntimeError("Vídeo não foi encontrado")

fps = video.get(cv2.CAP_PROP_FPS)

frames = int(video.
             get(cv2.CAP_PROP_FRAME_COUNT))

print("FPS: ", fps)
print("Número de frames: ", frames)
print("Duração: ", frames/fps, "s")
```
